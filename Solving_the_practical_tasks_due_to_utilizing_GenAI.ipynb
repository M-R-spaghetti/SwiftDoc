##Task 1:
When one atom of uranium-235 undergoes fission, it releases approximately 200 MeV of energy. Meanwhile, the combustion of 1 kg of coal releases approximately 24 MJ of thermal energy. How much energy (in Joules) is released be the complete fission of 1 gram of U-235 ? How many kilograms of coal would be needed to produce the same amount of energy ?

# ⚛️ Energy Comparison: Uranium Fission vs. Coal Combustion

## Explanation of Variables and Formulas

### Energy from Uranium Fission

The total energy released from uranium fission is calculated with the formula:

$$E_{U} = \left( \frac{m_{U}}{M_{U}} \right) \times N_{A} \times E_{\text{fission}} \times C$$

Where:
* **$m_{U}$ (mass of uranium)**: The mass of the Uranium-235 sample. In our case, this is **1 gram**.
* **$M_{U}$ (molar mass of uranium)**: The mass of one mole of Uranium-235 atoms, which is **235 g/mol**.
* **$E_{\text{fission}}$ (energy per fission)**: The energy released during the fission of a single atom. It is equal to **200 MeV**.
* **$E_{U}$ (total energy from uranium)**: The final energy output, measured in **Joules (J)**.

---

### Equivalent Mass of Coal

To find the mass of coal that will yield the same amount of energy, this formula is used:

$$m_{\text{coal}} = \frac{E_{U}}{E_{\text{coal}}}$$

Where:
* **$m_{\text{coal}}$ (mass of coal)**: The mass of coal that needs to be burned. It is measured in **kilograms (kg)**.
* **$E_{\text{coal}}$ (energy per kg of coal)**: The energy released when burning 1 kg of coal. It is equal to **24 MJ/kg**.

---

### Physical Constants and Coefficients

* **$N_{A}$ (Avogadro's constant)**: The number of atoms in one mole of a substance. Its value is **$6.022 \times 10^{23}$ atoms/mol**.
* **$C$ (conversion factor)**: The factor for converting energy from megaelectronvolts (MeV) to Joules (J). **1 MeV = $1.602 \times 10^{-13}$ J**.

```python
# Constants
mass_uranium_g = 1  # 1 gram of U-235
molar_mass_uranium = 235  # g/mol
avogadro_number = 6.022e23  # atoms/mol
energy_per_fission_MeV = 200  # MeV
MeV_to_J = 1.602e-13  # Joules per MeV
energy_per_kg_coal_MJ = 24  # MJ/kg
MJ_to_J = 1e6 # Joules per MJ

# 1. Calculate the number of atoms in 1 gram of U-235
moles_uranium = mass_uranium_g / molar_mass_uranium
number_of_atoms = moles_uranium * avogadro_number

# 2. Calculate the total energy released by the fission of 1 gram of U-235 in MeV
total_energy_MeV = number_of_atoms * energy_per_fission_MeV

# 3. Convert the energy from MeV to Joules
total_energy_J = total_energy_MeV * MeV_to_J

# 4. Calculate the kilograms of coal needed to produce the same amount of energy
energy_per_kg_coal_J = energy_per_kg_coal_MJ * MJ_to_J
coal_mass_kg = total_energy_J / energy_per_kg_coal_J

print(f"Energy released by 1 gram of U-235: {total_energy_J:.2e} J")
print(f"Kilograms of coal needed for the same energy: {coal_mass_kg:.2f} kg")
```

## Task 2
If the specific heat of iron is 450 J/(kg°C), how much energy is required to bring 0.5 kg of iron
from 0°C to 100°C?

### Formula for Heat Energy 🌡️

The formula to calculate the heat energy ($Q$) required to change the temperature of a substance is:

$$Q = m \cdot c \cdot \Delta T$$

* **$Q$ (Heat Energy):** This is the total amount of energy transferred to the substance, measured in **Joules (J)**.
* **$m$ (Mass):** This represents the mass of the substance being heated, measured in **kilograms (kg)**.
* **$c$ (Specific Heat Capacity):** This is an intrinsic property of the material. It's the amount of energy required to raise the temperature of 1 kg of the substance by 1°C. For iron, this value is **450 J/(kg°C)**.
* **$\Delta T$ (Change in Temperature):** This is the difference between the final and initial temperatures ($\Delta T = T_{final} - T_{initial}$), measured in **degrees Celsius (°C)**.

```python
# Define the given values
specific_heat_iron = 450  # J/(kg°C)
mass_iron = 0.5           # kg
initial_temp = 0          # °C
final_temp = 100          # °C

# Calculate the change in temperature
delta_t = final_temp - initial_temp

# Calculate the required energy using the formula Q = m * c * ΔT
energy_required = mass_iron * specific_heat_iron * delta_t

# Print the result
print(f"The energy required is: {int(energy_required)} Joules.")
```

# Mean Life and Specific Activity of Cobalt-60 (⁶⁰Co)

Cobalt-60 (⁶⁰Co), a synthetic radioactive isotope of cobalt, has a **mean lifetime of approximately 7.60 years** and a **specific activity of about 4.18 x 10¹³ Bq/g** (becquerels per gram). These fundamental properties dictate its behavior and applications in various fields, including medicine and industry. 🧪

---

## Mean Lifetime

The **mean lifetime (τ)** of a radionuclide is the average time a nucleus is expected to exist before it undergoes radioactive decay. It is mathematically related to the **half-life (t₁/₂)**, which is the time it takes for half of the radioactive nuclei in a sample to decay.

The half-life of Cobalt-60 is well-established as **5.27 years**. The mean lifetime is calculated using the following formula:

$$ \tau = \frac{t_{1/2}}{\ln(2)} $$

Given that ln(2) is approximately 0.693, the mean lifetime of ⁶⁰Co is:

$$ \tau = \frac{5.27 \text{ years}}{0.693} \approx 7.60 \text{ years} $$

This means that, on average, a Cobalt-60 atom will remain undecayed for about **7.60 years**.

---

## Specific Activity

The **specific activity (SA)** of a radioactive material is the activity per unit mass of the radionuclide. It is a measure of how radioactive a given amount of the substance is. The specific activity is determined by the decay constant (λ) and the molar mass (M) of the isotope.

The decay constant is inversely related to the mean lifetime:

$$ \lambda = \frac{1}{\tau} $$

First, we convert the mean lifetime from years to seconds:

$$ \tau \approx 7.60 \text{ years} \times 3.1536 \times 10^7 \text{ s/year} \approx 2.397 \times 10^8 \text{ s} $$

Therefore, the decay constant is:

$$ \lambda = \frac{1}{2.397 \times 10^8 \text{ s}} \approx 4.17 \times 10^{-9} \text{ s}^{-1} $$

The specific activity is then calculated using the formula:

$$ SA = \frac{\lambda \times N_A}{M} $$

Where:
* $N_A$ is Avogadro's number, approximately $6.022 \times 10^{23}$ atoms/mol.
* $M$ is the molar mass of Cobalt-60, which is approximately 59.93 g/mol.

Plugging in the values, we get:

$$ SA = \frac{(4.17 \times 10^{-9} \text{ s}^{-1}) \times (6.022 \times 10^{23} \text{ atoms/mol})}{59.93 \text{ g/mol}} \approx 4.19 \times 10^{13} \text{ atoms s}^{-1} \text{g}^{-1} $$

Since one becquerel (Bq) is equal to one decay per second, the specific activity of Cobalt-60 is approximately **4.18 x 10¹³ Bq/g**. This high level of radioactivity makes ⁶⁰Co a potent source of gamma radiation for medical radiotherapy and industrial applications such as sterilization and radiography.

```python
import numpy as np

# --- Constants for Cobalt-60 ---
half_life_years = 5.27  # Half-life of Co-60 in years
molar_mass_g_mol = 59.93  # Molar mass of Co-60 in g/mol
avogadro_number = 6.022e23  # Avogadro's number in atoms/mol

# --- Conversions ---
seconds_per_year = 3.1536e7 # Approximate number of seconds in a year

def calculate_mean_life(half_life):
  """Calculates the mean lifetime from the half-life."""
  return half_life / np.log(2)

def calculate_specific_activity(half_life_years, molar_mass):
  """Calculates the specific activity in Bq/g."""
  # Convert half-life to seconds
  half_life_seconds = half_life_years * seconds_per_year

  # Calculate the decay constant (lambda) in s^-1
  decay_constant = np.log(2) / half_life_seconds

  # Calculate specific activity: SA = (lambda * N_A) / M
  specific_activity = (decay_constant * avogadro_number) / molar_mass
  return specific_activity

# --- Perform Calculations ---
mean_life = calculate_mean_life(half_life_years)
specific_activity = calculate_specific_activity(half_life_years, molar_mass_g_mol)

# --- Display Results ---
print("--- Cobalt-60 (⁶⁰Co) Properties ---")
print(f"Mean Lifetime: {mean_life:.2f} years")
print(f"Specific Activity: {specific_activity:.2e} Bq/g")
```

## Formula for Calculating the Average Number of Collisions

The formula used to calculate the average number of collisions required to slow down a neutron is as follows:

$$ N = \frac{1}{\xi} \ln\left(\frac{E_{initial}}{E_{final}}\right) $$

Де:
*   $N$ is the average number of collisions.
*   $\xi$ (xi) is the average logarithmic energy decrement of the moderator. This value indicates how much energy a neutron loses on average in one collision with an atom of the moderator.
*   $E_{initial}$ is the initial energy of the neutron (in eV).
*   $E_{final}$ is the final (thermal) energy of the neutron (in eV).
*   $\ln$ is the natural logarithm.

This formula is based on the assumption that neutron moderation occurs as a result of elastic collisions with the nuclei of the moderator. The larger the value of ξ for a moderator, the fewer collisions are needed for a neutron to reach thermal energy.

## Calculation of the Average Logarithmic Energy Decrement ($\xi$)

The average logarithmic energy decrement ($\xi$) for the elastic scattering of neutrons on the nuclei of a moderator is calculated using the formula:

$$\xi = 1 + \frac{(A-1)^2}{2A} \ln\left(\frac{A-1}{A+1}\right)$$

Where $A$ is the mass number of the moderator nucleus.

This formula is derived from considering the kinematics of an elastic collision between a neutron and a nucleus. It shows that $\xi$ depends solely on the mass number of the moderator. The smaller the mass number $A$, the larger the value of $\xi$, meaning the more energy the neutron loses in a single collision.

For example, for hydrogen ($A=1$):
$$\xi_H = 1 + \frac{(1-1)^2}{2 \times 1} \ln\left(\frac{1-1}{1+1}\right) = 1 + 0 = 1$$
Although in reality, a slightly different formula or value is used for hydrogen, as at $A=1$ the denominator becomes zero in the original formula, and other factors are taken into account. However, the standard accepted value for hydrogen is $\xi \approx 1.0$.

For carbon ($A=12$):
$$\xi_C = 1 + \frac{(12-1)^2}{2 \times 12} \ln\left(\frac{12-1}{12+1}\right) = 1 + \frac{121}{24} \ln\left(\frac{11}{13}\right)$$
$$\xi_C = 1 + 5.04166 \times \ln(0.84615) \approx 0.158$$


## Detailed Description of Calculations

Here are the steps that were performed to calculate the average number of collisions for neutron moderation in hydrogen and carbon:

1.  **Defining Initial Data:**
    * Initial neutron energy (after fission): `initial_energy = 2e6` eV (2 MeV).

    * Average logarithmic energy decrement ($\xi$) for carbon: `xi_carbon = 0.158`.

    $$N = \frac{1}{\xi} \ln\left(\frac{E_{initial}}{E_{final}}\right)$$

        $$\ln\left(\frac{2 \times 10^6}{0.025}\right) \approx 18.206$$

4.  **Calculation for Carbon:**

        $$\ln\left(\frac{2 \times 10^6}{0.025}\right) \approx 18.206$$
    * I divided the result by $\xi$ for carbon:
        $$N_{carbon} = \frac{1}{0.158} \times 18.206 \approx 115.227$$

Thus, the provided formula was used, the corresponding values for the initial and final neutron energy were substituted, as well as the values of the average logarithmic energy decrement for each moderator ($\xi$), and the average number of collisions was calculated for each case. The obtained results were then compared.

```python
import math
```

```python
def calculate_collisions(e_initial, e_final, xi):
  """
  Calculates the average number of collisions for a neutron to slow down.

  Args:
    e_initial: The initial energy of the neutron in eV.
    e_final: The final (thermal) energy of the neutron in eV.
    xi: The average logarithmic energy decrement of the moderator.

  Returns:
    The average number of collisions.
  """
  return (1 / xi) * math.log(e_initial / e_final)

# --- Define Constants ---

# Initial fission neutron energy in eV
initial_energy = 2e6  # 2 MeV

# Final thermal energy in eV
final_energy = 0.025

# Average logarithmic energy decrement (xi) for different materials
# These are standard, accepted values in nuclear engineering.
xi_hydrogen = 1.0
xi_carbon = 0.158

# --- Perform Calculations ---

collisions_hydrogen = calculate_collisions(initial_energy, final_energy, xi_hydrogen)
collisions_carbon = calculate_collisions(initial_energy, final_energy, xi_carbon)

# --- Display Results ---

print("--- Neutron Moderation Comparison ---")
print(f"Initial Energy: {initial_energy / 1e6} MeV")
print(f"Final Thermal Energy: {final_energy} eV\n")

print(f"Moderator: Hydrogen (H)")
print(f"Average Logarithmic Energy Decrement (ξ): {xi_hydrogen}")
print(f"Average number of collisions to thermalize: {collisions_hydrogen:.2f}\n")

print(f"Moderator: Carbon (C)")
print(f"Average Logarithmic Energy Decrement (ξ): {xi_carbon}")
print(f"Average number of collisions to thermalize: {collisions_carbon:.2f}\n")

print("--- Conclusion ---")
print(f"It takes approximately {collisions_carbon / collisions_hydrogen:.1f} times more collisions to thermalize a neutron in Carbon than in Hydrogen.")
print("This is because hydrogen's mass is very close to the neutron's mass, allowing for a much larger energy transfer per collision.")
```
