## ☢️ Задача 1 / Task 1: Порівняння енергії поділу Урану-235 та горіння Вугілля

### **Умова:**
При поділі одного атома урану-235 виділяється приблизно **200 МеВ** енергії. При згорянні 1 кг вугілля виділяється **24 МДж** теплової енергії.
1. Скільки енергії (у Джоулях) виділяється при повному поділі **1 грама U-235**?
2. Скільки кілограмів вугілля знадобиться для отримання такої ж кількості енергії?

### **Фізична формула:**
$$E_{U} = \left( \frac{m_{U}}{M_{U}} \right) \times N_{A} \times E_{\text{fission}} \times 1.602 \times 10^{-13}$$
$$m_{\text{coal}} = \frac{E_{U}}{q_{\text{coal}}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Константи
m_u = 1.0  # маси урану в грамах
M_u = 235.0  # молярна маса г/моль
N_A = 6.022e23  # число Авогадро
E_fission_MeV = 200.0  # МеВ на один поділ
MeV_to_J = 1.602e-13  # Джоулів в 1 МеВ
q_coal_MJ = 24.0  # МДж/кг вугілля
MJ_to_J = 1e6

# Розрахунок
N_atoms = (m_u / M_u) * N_A
E_uranium_J = N_atoms * E_fission_MeV * MeV_to_J
m_coal_kg = E_uranium_J / (q_coal_MJ * MJ_to_J)

print("="*60)
print(f"1. Енергія з 1г U-235: {E_uranium_J:.3e} Джоулів ({E_uranium_J/1e9:.2f} ГДж)")
print(f"2. Маса вугілля для еквіваленту: {m_coal_kg:.2f} кг ({m_coal_kg/1000:.3f} тонн)")
print("="*60)

# Візуалізація
plt.figure(figsize=(7, 4))
plt.bar(["1 грам Урану-235", f"{m_coal_kg:.0f} кг Вугілля (~3.4 т)"], [E_uranium_J/1e9, (m_coal_kg * q_coal_MJ * MJ_to_J)/1e9], color=["limegreen", "dimgray"])
plt.ylabel("Виділена Енергія (ГДж)")
plt.title("Еквівалентність енергії: 1г U-235 vs Вугілля")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

---## 🌡️ Задача 2 / Task 2: Тепловий розрахунок нагріву Заліза

### **Умова:**
Питома теплоємність заліза становить **450 Дж/(кг·°C)**. Скільки енергії знадобиться, щоб нагріти **0.5 кг заліза** від **0°C до 100°C**?

### **Фізична формула:**
$$Q = m \cdot c \cdot \Delta T = m \cdot c \cdot (T_{\text{final}} - T_{\text{initial}})$$

```python
c_iron = 450.0  # Дж/(кг·°C)
m_iron = 0.5    # кг
T_init = 0.0    # °C
T_final = 100.0 # °C

delta_T = T_final - T_init
Q_joules = m_iron * c_iron * delta_T

print("="*60)
print(f"Необхідна теплова енергія Q = {Q_joules:.1f} Дж ({Q_joules/1000:.2f} кДж)")
print("="*60)

# Графік нагріву
temps = np.linspace(T_init, T_final, 50)
energy_curve = m_iron * c_iron * (temps - T_init)

plt.figure(figsize=(7, 4))
plt.plot(temps, energy_curve / 1000, color="crimson", lw=2.5)
plt.xlabel("Температура (°C)")
plt.ylabel("Поглинута енергія (кДж)")
plt.title("Залежність енергії нагріву 0.5 кг заліза від температури")
plt.grid(True)
plt.show()
```

---## 💎 Задача 3 / Task 3: Концентрація ядер в Графітовому Замедлителі

### **Умова:**
Обчислити кількість атома вуглецю-12 (C-12) в **1 см³ графиту** з густиною **1.65 г/см³**.

### **Фізична формула:**
$$N_{\text{atoms}} = \left( \frac{V \cdot \rho}{M} \right) \cdot N_{A}$$

```python
rho_graphite = 1.65  # г/см³
vol_graphite = 1.0   # см³
M_c12 = 12.01        # г/моль
N_A = 6.022e23       # атоми/моль

mass_g = rho_graphite * vol_graphite
moles = mass_g / M_c12
atoms_count = moles * N_A

print("="*60)
print(f"Кількість атомів C-12 в 1 см³ графіту: {atoms_count:.3e} атомів")
print("="*60)
```

---## 🧪 Задача 4 / Task 4: Період напіврозпаду та Активність Кобальту-60 (Co-60)

### **Умова:**
Визначити середній час життя (Mean Life) та питому активність (Specific Activity) для ізотопу **Кобальт-60 (Co-60)** ($T_{1/2} = 5.27$ років).

### **Фізичні формули:**
$$\tau = \frac{T_{1/2}}{\ln 2}$$
$$A_{\text{spec}} = \frac{\lambda \cdot N_{A}}{M} = \frac{\ln 2}{T_{1/2} \cdot M} \cdot N_{A}$$

```python
T_half_years = 5.27
sec_per_year = 3.1536e7
T_half_sec = T_half_years * sec_per_year
M_co60 = 59.93  # г/моль
N_A = 6.022e23

tau_years = T_half_years / np.log(2)
lam_sec = np.log(2) / T_half_sec
spec_activity_Bq_g = (lam_sec * N_A) / M_co60

print("="*60)
print(f"1. Середній час життя (Mean Life, tau): {tau_years:.2f} років")
print(f"2. Питома активність Co-60: {spec_activity_Bq_g:.3e} Бк/г ({spec_activity_Bq_g/3.7e10:.2f} Кюрі/г)")
print("="*60)

# Графік радіоактивного розпаду
years = np.linspace(0, 25, 100)
remaining_fraction = np.exp(-years / tau_years)

plt.figure(figsize=(7, 4))
plt.plot(years, remaining_fraction * 100, color="purple", lw=2.5)
plt.axvline(T_half_years, color="orange", linestyle="--", label=f"T1/2 = {T_half_years} років")
plt.xlabel("Роки")
plt.ylabel("Залишок ізотопу Co-60 (%)")
plt.title("Закон радіоактивного розпаду Кобальту-60")
plt.legend()
plt.grid(True)
plt.show()
```

---## 🛡️ Задача 5 / Task 5: Замедлення Нейтронів (Водород vs Углерод)

### **Умова:**
Порівняти кількість зіткнень, необхідних для уповільнення швидкого нейтрона від енергії **2 МеВ** до теплової енергії **0.025 еВ** у водороді ($H$) та углероді ($C$).

### **Фізична формула:**
$$N = \frac{1}{\xi} \ln\left(\frac{E_{\text{initial}}}{E_{\text{final}}}\right)$$
де $\xi$ — середній логарифмічний декремент загасання енергії ($\xi_H = 1.0$, $\xi_C = 0.158$).

```python
E_init = 2e6   # 2 МеВ в еВ
E_final = 0.025 # 0.025 еВ
xi_H = 1.0
xi_C = 0.158

N_H = (1 / xi_H) * np.log(E_init / E_final)
N_C = (1 / xi_C) * np.log(E_init / E_final)
ratio = N_C / N_H

print("="*60)
print(f"1. Зіткнень у Водороді (H): {N_H:.2f}")
print(f"2. Зіткнень у Вуглеці (C): {N_C:.2f}")
print(f"3. Вуглець потребує в {ratio:.1f} разів більше зіткнень, ніж Водород!")
print("="*60)

# Візуалізація уповільнення
plt.figure(figsize=(7, 4))
plt.bar(["Водород (H)", "Вуглець (C)"], [N_H, N_C], color=["skyblue", "orange"])
plt.ylabel("Середня кількість зіткнень (N)")
plt.title("Ефективність уповільнення нейтронів (від 2 МеВ до 0.025 еВ)")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
